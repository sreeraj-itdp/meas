```python
#adlsAccountName = "measmmdevst01"
adlsAccountNameMeas = dbutils.secrets.get(scope = adb_scope, key = "ADLS-Account-Name")
adlsClientIdMeas = dbutils.secrets.get(scope = adb_scope, key = "Service-Principal-Client-ID")
adlsTenantIdMeas = dbutils.secrets.get(scope = adb_scope, key = "Service-Principal-Tenant-Id")
adlsSecretMeas = dbutils.secrets.get(scope = adb_scope, key = "Service-Principal-Secret")
adlsContainerName = "meas"
mountPoint = "/mnt/meas-"+env+""
source = "abfss://" + adlsContainerName + "@" + adlsAccountNameMeas + ".dfs.core.windows.net/"
 
configs = {"fs.azure.account.auth.type": "OAuth",
          "fs.azure.account.oauth.provider.type": "org.apache.hadoop.fs.azurebfs.oauth2.ClientCredsTokenProvider",
          "fs.azure.account.oauth2.client.id": adlsClientIdMeas,
          "fs.azure.account.oauth2.client.secret": adlsSecretMeas,
          "fs.azure.account.oauth2.client.endpoint": "https://login.microsoftonline.com/"+adlsTenantIdMeas+"/oauth2/token"}

if not any(mount.mountPoint == mountPoint for mount in dbutils.fs.mounts()):

  dbutils.fs.mount(
    source = source,
    mount_point = mountPoint,
    extra_configs = configs)
  print("Mounted!")
else:
  print("Mount point already exists!")
```

```python
adlsAccountNameitds = dbutils.secrets.get(scope = adb_scope, key = "itds-account-name")
adlsaccesskeyitds = dbutils.secrets.get(scope = adb_scope, key = "itds-blob-access-key")
if env == "qa1":
    itdsenv = "qa"
else:
    itdsenv = env
adlsContainerNameitds = "itds-"+itdsenv+"-direct-feeds"
itds_mountPoint = "/mnt/itds-"+itdsenv+""

if not any(mount.mountPoint == itds_mountPoint for mount in dbutils.fs.mounts()):
    dbutils.fs.mount(
        source = "wasbs://" + adlsContainerNameitds + "@" + adlsAccountNameitds + ".blob.core.windows.net",
        mount_point = "/mnt/itds-"+itdsenv+"",
        extra_configs = {"fs.azure.account.key." + adlsAccountNameitds + ".blob.core.windows.net": adlsaccesskeyitds})
    print("Mounted!")
else:
    print("Mount point already exists!")
```
